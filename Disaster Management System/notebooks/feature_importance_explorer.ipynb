# 📊 Interactive Feature Importance Explorer (PyGWalker)
Explore XGBoost v3 feature importances interactively. Filter, sort, and visualize key predictors.

```python
import os
import joblib
import pandas as pd
import numpy as np
import pygwalker as pyg

# Load trained XGBoost v3 model artifact
model_path = '../models/xgboost_flood_model_v3.pkl'
model_data = joblib.load(model_path)
model = model_data['model']
feature_names = model_data['feature_names']
importances = model.feature_importances_

# Create Feature Importance DataFrame
importance_df = pd.DataFrame({
    'feature': feature_names,
    'importance_score': np.round(importances, 4),
    'percentage': np.round(importances * 100, 2)
}).sort_values(by='importance_score', ascending=False).reset_index(drop=True)

# Display DataFrame Summary
print('Top Feature Importance Scores:')
display(importance_df)

# Launch Interactive PyGWalker GUI
walker = pyg.walk(importance_df)
```
